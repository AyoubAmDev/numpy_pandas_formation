```python
import pandas as pd
import numpy as np

print("=== Loading Data ===")
df = pd.read_csv('club_members.csv')
print("Data loaded successfully!")
```

```python
print("\n=== First Look ===")
print("First 3 rows:")
print(df.head(3))

print("___________________________________________")

print("\nDataFrame shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nBasic info:")
df.info()
```

```python
# Selecting data
print("\n=== Selecting Data ===")
print("Just the names:")
print(df['Name'])

print("--------------------------")

print("\nMultiple columns:")
print(df[['Name', 'Major']])

print("--------------------------")

print("\nFirst 2 rows, specific columns:")
print(df.loc[0:1, ['Name', 'Major', 'Year']])

print("--------------------------")

print("\nFiltering - Python lovers:")
python_lovers = df[df['Favorite_Programming_Language'] == 'Python']
print(python_lovers)
```

```python
# Cell 4: Basic operations and statistics
print("\n=== Statistics & Operations ===")
print("Descriptive stats:")
print(df.describe())

print("\nAverage coffee consumption:", df['Coffee_Cups_Per_Day'].mean())
print("Most common major:", df['Major'].mode()[0])

print("\nValue counts for majors:")
print(df['Major'].value_counts())
```

```python
# Adding and modifying columns
print("\n=== Creating New Columns ===")
df['Productivity_Score'] = df['Projects_Completed'] / df['Coffee_Cups_Per_Day']
df['Is_Senior'] = df['Year'] >= 3

print("DataFrame with new columns:")
print(df[['Name', 'Productivity_Score', 'Is_Senior']])
```

```python
# Grouping and aggregation (POWER FEATURE!)
print("\n=== Grouping Data ===")
print("Average projects completed by major:")
major_stats = df.groupby('Major').agg({
    'Projects_Completed': 'mean',
    'Coffee_Cups_Per_Day': 'mean',
    'Name': 'count'  # Count of students
}).rename(columns={'Name': 'Student_Count'})

print(major_stats)
```

```python
# Handling missing data (quick demo)
print("\n=== Handling Missing Data ===")
# Let's intentionally create some missing data
df_with_na = df.copy()
df_with_na.loc[0, 'Projects_Completed'] = np.nan

print("Data with missing values:")
print(df_with_na.isna().sum())

print(df_with_na.head(2))

print("\nFilled missing values:")
df_filled = df_with_na.fillna({'Projects_Completed': 0})
print(df_filled.head(2))
```
